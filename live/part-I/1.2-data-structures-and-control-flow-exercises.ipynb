````python
numbers= [6.1, 7.4, 5.9, 8.2, 6.8]
def mean(values):
  "return the mean of a list of values"
  total = 0.0
  for v in values:
    total+= v
  return total/len(values)
print(round(mean(numbers),2))
````

## Exercise 10: Fix the shared-default bug

An assistant wrote this function; it is correct the first time but wrong on later calls:

```python
def collect_above(values, threshold, out=[]):

        if v > threshold:
            out.append(v)
    return out
```

In one comment, explain the bug, then rewrite it so each call starts fresh. Demonstrate with two
separate calls, assigning each result to its own variable.

````python
# The default list is shared between calls if we use out=[], so using None lets us create a fresh list for each call.
def collect_above(values, threshold, out=None):
    if out is None:
        out = []

    for v in values:
        if v > threshold:
            out.append(v)

    return out
result_1 = collect_above([6.1, 7.4, 5.9, 8.2], 7.0)
result_2 = collect_above([3.2, 9.1, 4.5, 10.2], 8.0)

print(result_1)
print(result_2)
````

This function is meant to return the readings without the last one:

### Exercise 15: Reject impossible temperatures

Write `to_kelvin(temp_celsius)` that raises a `ValueError` when the input is below absolute zero
(−273.15 °C) and otherwise returns the temperature in kelvin. Give it a docstring and a type hint.

Loop over `[20.0, -300.0, 0.0]`, calling it inside a `try`/`except` so that the invalid value is
reported without stopping the loop.

````python
def to_kelvin(temp_celsius: float) -> float:
    """Convert Celsius to Kelvin and reject temperatures below absolute zero."""
    if temp_celsius < -273.15:
        raise ValueError("Temperature cannot be below absolute zero.")
    return temp_celsius + 273.15


for temp in [20.0, -300.0, 0.0]:
    try:
        kelvin = to_kelvin(temp)
        print(f"{temp} °C = {kelvin} K")
    except ValueError as error:
        print(f"Invalid temperature: {temp} °C — {error}")
````

**Step 1.** Write a function `fahrenheit_to_celsius(temp_fahrenheit)` that returns the
temperature in °C. Test it on the freezing point of water (32 °F should give 0.0 °C) and on
212 °F. A one-line docstring is optional here (see Exercise 11).

You wrote this conversion inline in 1.1. Steps 2 and 4 both need it again; with a function,
they call the same tested code instead of repeating the arithmetic.

````python
def fahrenheit_to_celsius(temp_fahrenheit):
    """Convert a temperature from Fahrenheit to Celsius."""
    return (temp_fahrenheit - 32) * 5 / 9

print(fahrenheit_to_celsius(32))
print(fahrenheit_to_celsius(212))
````

**Step 2.** Open the file, read all lines, and loop over the records (skipping the header).
For each one, extract the month and the temperature in °C. Print the first three as a check.

Sixteen days in this file have an empty temperature field — a real gap in the station's record,
late October to mid-November. `float("")` raises `ValueError`, so your loop has to decide what to
do with those days before it can get to the end of the file. Skip them, and count how many you
skipped. After the loop, print how many records you converted and how many you skipped; the two
should add up to 365.

````python
from pathlib import Path

skipped = 0
converted = 0

with Path(data_file).open("r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

for line in lines[1:]:
    fields = line.strip().split(",")

    month = fields[1].split(".")[1]

    if fields[2] == "":
        skipped += 1
        continue

    temp_fahrenheit = float(fields[2])
    temp_celsius = fahrenheit_to_celsius(temp_fahrenheit)

    converted += 1

    if converted <= 3:
        print(month, temp_celsius)

print("converted:", converted)
print("skipped:", skipped)
print("total:", converted + skipped)
````

**Step 3.** Build a dictionary mapping each month to the number of observations in it.

You do not know in advance which months are present, so the dictionary has to grow as you go.
Keep skipping the days with no reading: a day the sensor did not record is not an observation, so
October and November should come out short of their full length.

````python
counts = {}

with Path(data_file).open("r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

for line in lines[1:]:
    fields = line.strip().split(",")

    if fields[2] == "":
        continue

    month = fields[1].split(".")[1]

    counts[month] = counts.get(month, 0) + 1

print(counts)
````

**Step 4.** Build a second dictionary mapping each month to its **mean** daily maximum
temperature in °C. Print it with each mean rounded to two decimals.

````python
counts = {}
totals = {}

with Path(data_file).open("r", encoding="utf-8") as fhandler:
    lines = fhandler.readlines()

for line in lines[1:]:
    fields = line.strip().split(",")

    if fields[2] == "":
        continue

    month = fields[1].split(".")[1]

    temp_fahrenheit = float(fields[2])
    temp_celsius = fahrenheit_to_celsius(temp_fahrenheit)

    counts[month] = counts.get(month, 0) + 1
    totals[month] = totals.get(month, 0) + temp_celsius

means = {}

for month in totals:
    means[month] = totals[month] / counts[month]

print(means)
````

**Step 5.** Find the warmest month and print it with its mean temperature.

Do this with a loop and a comparison, keeping track of the best value seen so far, as you did for
the warmest day in Exercise 6. A dictionary is looked up by key, not by position, so there is no
`[0]` to start from: start from `None` instead.

````python
warmest_month = None
warmest_mean = None

for month, mean_temp in means.items():
    if warmest_mean is None:
        warmest_month = month
        warmest_mean = mean_temp
    elif mean_temp > warmest_mean:
        warmest_month = month
        warmest_mean = mean_temp

print("warmest month:", warmest_month)
print("mean temperature:", round(warmest_mean, 2))
````

**Step 6.** Write your monthly summary to `_files/monthly_means.csv`, with the header
`month,n_obs,mean_temp_celsius` and one line per month, creating the `_files` folder first as you
did in 1.1. Then read the file back and print it, to confirm it says what you meant it to say.

````python
summary_path = Path("_files/monthly_means.csv")
summary_path.parent.mkdir(exist_ok=True)

with summary_path.open("w", encoding="utf-8") as fhandler:
    fhandler.write("month,n_obs,mean_temp_celsius\n")

    for month in means:
        fhandler.write(
            f"{month},{counts[month]},{means[month]:.2f}\n"
        )

with summary_path.open("r", encoding="utf-8") as fhandler:
    summary = fhandler.read()

print(summary)
````

````python
def to_planet_masses(mass_1e24kg, reference="jupiter"):
    return mass_1e24kg / planet_masses_1e24kg[reference]


jupiter_earth = to_planet_masses(
    planet_masses_1e24kg["jupiter"],
    reference="earth"
)

jupiter_jupiter = to_planet_masses(
    planet_masses_1e24kg["jupiter"],
    reference="jupiter"
)

print(jupiter_earth)
print(jupiter_jupiter)
````

**Q11) Write a function that returns two values.**

`mass_in_two_units(mass_1e24kg, reference)` should return the mass in Earth masses *and* in masses
of whatever reference planet you pass. Test it on Jupiter with `reference="mars"`.

Hint: `return a, b` returns a tuple, and `first, second = f(...)` unpacks it — both from 1.2's
lecture.

````python
def mass_in_two_units(mass_1e24kg, reference):
    earth_masses = mass_1e24kg / planet_masses_1e24kg["earth"]
    reference_masses = mass_1e24kg / planet_masses_1e24kg[reference]

    return earth_masses, reference_masses
earth_masses, mars_masses = mass_in_two_units(
    planet_masses_1e24kg["jupiter"],
    reference="mars"
)

print(earth_masses)
print(mars_masses)
````
