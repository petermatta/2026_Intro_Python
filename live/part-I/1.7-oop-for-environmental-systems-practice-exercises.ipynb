## Exercise 1: A class with state and behaviour

Define a `Glacier` class with `name` and `area_km2` instance attributes, a `describe()` method returning a short string, and a `__repr__`. Create one and print both the object and its description.

````python
# Define the Glacier class
class Glacier:

    # Initialize the object with its name and area
    def __init__(self, name, area_km2):
        self.name = name
        self.area_km2 = area_km2

    # Return a short description of the glacier
    def describe(self):
        return f"{self.name} covers {self.area_km2} km²."

    # Define how the object is displayed when printed
    def __repr__(self):
        return f"Glacier(name='{self.name}', area_km2={self.area_km2})"


# Create a Glacier object
glacier1 = Glacier("Meow Glacier", 78.0)

# Print the object itself
print(glacier1)

# Print the glacier's description
print(glacier1.describe())
````

## Exercise 2: Instance versus class attributes

Define a `Planet` class with a shared class attribute `g_earth = 9.81` and per-instance attributes `name` and `surface_gravity`. Create two planets and show that `g_earth` is shared while `surface_gravity` differs.

````python
# Define the Planet class
class Planet:

    # Class attribute: shared by all Planet objects
    g_earth = 9.81

    def __init__(self,name,surface_gravity):
      self.name = name
      self.surface_gravity = surface_gravity

    def _repr_(self):
      return f"Planet (name= '{self.name}', of surface_gravity= {self.surface_gravity})"

Planet1= Planet("Potato", 10.89)
Planet2= Planet("Nuggetosaurus", 13.81)

# Print the shared class attribute
print("Is the g-force of Potato planet equal that of Nuggetosaurus planet?",Planet1.g_earth==Planet2.g_earth)
print("Is the surface gravity of Potato planet equal that of Nuggetosaurus planet?", Planet1.surface_gravity == Planet2.surface_gravity)
#We can thus see that the g-force is shared while the surface gravity isnt.
````

## Exercise 4: Read and mutate methods, with validation

Define a `Measurements` class holding a list of floats, with `add(x)` to append and `mean()` to return the average. `mean()` should raise `ValueError` if there are no values. Demonstrate both.

````python
# Define the Measurements class
class Measurements:

    # Initialize the object with an empty list of values
    def __init__(self):
        self.values = []

    # Add a new measurement to the list
    def add(self, x):
        self.values.append(x)

    def mean(self):
        # Raise an error if the list is empty
        if len(self.values) == 0:
            raise ValueError("Cannot calculate the mean of an empty list")

        return sum(self.values) / len(self.values)
# Create a Measurements object
m = Measurements()

# Demonstrate the error when there are no measurements
try:
    print(m.mean())
except ValueError as e:
    print(e)

# Add measurements
m.add(10.0)
m.add(20.0)
m.add(30.0)

# Print the list and its mean
print(m.values)
print(m.mean())
````

## Exercise 5: Composition

Define a `Catchment` class that holds several `Measurements` objects keyed by name (composition). Add an `add_series(name, series)` method and a `names()` method returning the keys. Build one with two series.

````python
# Define the Catchment class
class Catchment:

    # Initialize an empty dictionary to store Measurements objects
    def __init__(self):
        self.series = {}

    # Add a Measurements object under a given name
    def add_series(self, name, series):
        self.series[name] = series

    # Return the names of all stored series
    def names(self):
        return self.series.keys()

# Create two Measurements objects
rainfall = Measurements()
rainfall.add(10.0)
rainfall.add(15.0)
rainfall.add(20.0)

temperature = Measurements()
temperature.add(12.0)
temperature.add(14.0)
temperature.add(16.0)

# Create a Catchment object
catchment2 = Catchment()

# Add both series to the catchment
catchment2.add_series("rainfall", rainfall)
catchment2.add_series("temperature", temperature)

# Print the names of the stored series
print(catchment2.names())
````
